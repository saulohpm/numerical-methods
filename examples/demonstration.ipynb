# Demonstration

Simple use of `numerical_methods` library.

## Cell 1 – Imports

```python
import numpy as np

# Imported library already installed with pip.
import numerical_methods as nm
```

## Cell 2 – Problem setup

```python
# Integration Interval
a = -np.pi
b = np.pi
L = (b - a) / 2
C = (a + b) / 2

# Interval Subdivisions or probability
n = 12

# Domain
nx = 256
x = np.linspace(a, b, nx)

# Step
delta_x = 0.3

# Function
f = lambda x: np.cos(x) ** 2 + np.sin(2 * x)
df_dx_exact = lambda x: -np.sin(2 * x) + 2*np.cos(2 * x)
d10f_dx10_exact = lambda x: -512 * np.cos(2 * x) - 1024 * np.sin(2 * x)
```

## Cell 3 – Numerical Integration

```python
rectangle = nm.rectangle_integrate(f, a, b, n)
trapezoid = nm.trapezoidal_integrate(f, a, b, n)

print(f"{'Rectangle Rule':30}: {rectangle:.6f}")
print(f"{'Trapezoid Rule':30}: {trapezoid:.6f}")
```

## Cell 4 – Numerical Differentiation

```python
evaluation_point = 2

central = nm.fd_central_derivative(f, evaluation_point, delta_x)
central_nth = nm.fd_nth_derivative(f, evaluation_point, delta_x)
richardson = nm.richardson_derivative(f, evaluation_point, delta_x)

print(f"{'Central Difference':30}: {central:.6f}")
print(f"{'Central Difference nth':30}: {central_nth:.6f}")
print(f"{'Richardson Method':30}: {richardson:.6f}")
```

## Cell 5 – Series Approximation

```python
fourier_approximation = nm.fourier_approx(f, x, L, n)
taylor_approximation = nm.taylor_approx(f, x, C, n)
```

## Cell 6 - Numerical Differentiation Function

```python
df_dx = nm.function_derivative(f, "central", delta_x)
d10f_dx10 = nm.function_derivative(f, "central_nth", delta_x, 10)
```

## Cell 7 – Graphs

```python
nm.plot_function(f, x)
nm.plot_function(f, x, a, b)
nm.plot_function(f, x, a, b, fourier_approximation)
nm.plot_function(f, x, a, b, taylor_approximation)
nm.plot_function(df_dx_exact, x, a, b, df_dx(x))
nm.plot_function(d10f_dx10_exact, x, a, b, d10f_dx10(x))
```
